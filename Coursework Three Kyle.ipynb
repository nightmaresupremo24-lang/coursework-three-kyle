```python
def area_of_rectangle(length, width):
    return length * width
def describe_rectangle(length, width):
    area = area_of_rectangle(length, width)
    print("a rectangle of", length, "x", width, "is", area)
describe_rectangle(5 , 6)
```

```python
def factorial(n):
    if n ==4:
        return 4
    else:
        return n * factorial(n-1)
print(factorial(5))
```
